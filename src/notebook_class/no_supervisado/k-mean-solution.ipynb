```python
import pandas as pd
import numpy as np
import random
```

```python
X = [[random.random(), random.random(), random.random()] for i in range(200)] 
X = np.array(X)
```

```python
#dt_l = list(zip([random.randint(1,5) for i in range(200)],X))
k = 5
max_it = 200
col_names = [f"c{i}" for i in range(X.shape[1])]
df = pd.DataFrame(X, columns=col_names)
df["label"] = [random.randint(1,k) for i in range(200)]
eq_labels = False
it = 0
while not eq_labels and it < max_it:
    it += 1
    mu_k = df.groupby("label").agg("mean")
    new_label = []
    for i_x in range(200):
        add_dist = []
        for k_i in range(k):
            add_dist.append(np.linalg.norm(np.array(df[col_names].values[i_x,:]) - np.array(mu_k.values[k_i,:])))
        new_label.append(np.argmin(add_dist))
    eq_labels = all(df["label"] == new_label)
    df["label"] = new_label
    print(f"it {it}, y eq labels are {eq_labels}")
df["label"]
mu_k
```

```python
class K_means:
    mu_k = None
    k = 3
    col_names = None
    def __init__(self, k, max_it=200):
        self.k = k
        self.max_it = max_it
    # x 200,3
    def fit(self, X):
        max_it = 200
        self.col_names = [f"c{i}" for i in range(X.shape[1])]
        df = pd.DataFrame(X, columns=col_names)
        df["label"] = [random.randint(0,self.k-1) for i in range(X.shape[0])]
        eq_labels = False
        it = 0
        while not eq_labels and it < max_it:
            it += 1
            self.mu_k = df.groupby("label").agg("mean").values
            new_label = self.predict(df[self.col_names])
            eq_labels = all(df["label"] == new_label)
            df["label"] = new_label
            print(f"it {it}, y eq labels are {eq_labels}")
    def predict(self, X):
        new_label = []
        for i_x in range(X.shape[0]):
            add_dist = []
            for k_i in range(self.k):
                add_dist.append(np.linalg.norm(np.array(X.values[i_x,:]) - 
                                               np.array(self.mu_k[k_i,:])))
            new_label.append(np.argmin(add_dist))
        return new_label
```

```python
kmeans = K_means(5)
```

```python
kmeans.fit(X)
```
